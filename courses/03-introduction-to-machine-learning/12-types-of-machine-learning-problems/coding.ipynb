Part 1: Classification (Predicting a Category)

```python
# 1. Classification: Predict if an email is Spam (1) or Not Spam (0)
from sklearn.tree import DecisionTreeClassifier

# Features (X): [Contains word 'free' (1/0), Number of links]
X_train = [[1, 5], [1, 8], [0, 1], [0, 0]]
# Labels (y): The answer key (Supervised)
y_train = [1, 1, 0, 0]  # 1 = Spam, 0 = Regular

# Step 1 & 2: Create model and learn rules from data + answers
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

# Step 3: Predict a brand new email [Contains 'free', 3 links]
new_email = [[1, 3]]
prediction = clf.predict(new_email)
print(
    f"Classification Result: {'SPAM' if prediction[0] == 1 else 'NOT SPAM'}"
)
```

Part 2: Regression (Predicting a Continuous Number)

```python
# 2. Regression: Predict apartment price based on area (sq meters)
from sklearn.linear_model import LinearRegression

# Features (X): [Area in sq meters]
X_train = [[45], [60], [80], [100]]
# Labels (y): Continuous numeric prices (in thousands $)
y_train = [50, 68, 92, 115]

# Step 1 & 2: Fit the line (dial) to data
reg = LinearRegression()
reg.fit(X_train, y_train)

# Step 3: Predict price for a 75 sq meter apartment
new_apartment = [[75]]
predicted_price = reg.predict(new_apartment)
print(f"Regression Result: Estimated price is ${predicted_price[0]:.2f}k")
```


Part 3: Clustering (Grouping Without Any Labels)

```python
# 3. Clustering: Group shoppers into 2 groups based on [Age, Spending Score]
from sklearn.cluster import KMeans

# Features only (X): Notice there is NO y_train here!
X_customers = [[19, 15], [22, 20], [45, 80], [50, 85]]

# Step 1 & 2: Discover 2 natural clusters with NO answer key
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X_customers)  # Notice: Only X is passed, no y!

# Check which cluster each customer was assigned to
print(f"Discovered Cluster IDs: {kmeans.labels_}")
```
